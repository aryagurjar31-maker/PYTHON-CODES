```python
#Reverse a String (Common DSA Task)
string = "Python"
reversed_str = ""
for ch in string:
  reversed_str = ch + reversed_str
print("Reversed:", reversed_str)
```

```python
#Counting Frequency of Elements (DSA Hash Map Use)
arr = [1, 2, 2, 3, 1, 4, 2]
freq = {}
for num in arr:
  freq[num] = freq.get(num, 0) + 1
print(freq)
```

```python
#Linear Search (Fundamental DSA Algorithm)
arr = [4, 7, 1, 9, 3]
key = 9
found = False
for val in arr:
  if val == key:
    found = True
    break

print("Element Found!" if found else "Not Found")
```

```python
#Factorial Using For Loop
n = 5
fact = 1
for i in range(1, n+1):
  fact *= i
print("Factorial of", n, "is", fact)
```

```python
#Right-Angled Triangle Pattern
# Program 1: Right-Angled Triangle
for i in range(1, 6):
    for j in range(1, i + 1):
      print("*", end=" ")
    print()
```

```python
#Inverted Right-Angled Triangle
# Program 2: Inverted Triangle

for i in range(5, 0, -1):
  for j in range(1, i + 1):
    print("*", end=" ")
  print()
```

```python
#Number Triangle Pattern
# Program 3: Number Pattern
for i in range(1, 6):
  for j in range(1, i + 1):
    print(j, end=" ")
  print()
```

```python
#Pyramid Pattern
# Program 4: Pyramid Pattern
rows = 5
for i in range(1, rows + 1):
  print(" " * (rows - i), end="") # Spaces
  print("* " * i)
```

```python
ch = 65 # ASCII for 'A'
for i in range(5):
    for j in range(i + 1):
        print(chr(ch), end=" ")
    ch += 1
    print()
```

```python
#Floyd’s Triangle (Number Sequence)
# Program 7: Floyd's Triangle
num = 1
for i in range(1, 6):
  for j in range(1, i + 1):
    print(num, end=" ")
    num += 1
  print()
```

```python
#Diamond Pattern
# Program 8: Diamond Pattern
rows = 5
# Upper part
for i in range(1, rows + 1):
  print(" " * (rows - i) + "* " * i)
# Lower part
for i in range(rows - 1, 0, -1):
  print(" " * (rows - i) + "* " * i)
```
