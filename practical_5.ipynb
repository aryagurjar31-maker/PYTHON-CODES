```python
def calculate(a, b):
    total = a + b
    difference = a - b
    product = a * b
    return total, difference, product

x, y, z = calculate(20, 10)

print("Total:", x)
print("Difference:", y)
print("Product:", z)
```

```python
def sum_all(*args):
    result = 0
    for num in args:
        result += num
    return result

print(sum_all(1, 2, 3, 4, 5))
```

```python
def display_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

display_info(name="Rahul", age=20, city="Pune")
```

```python
def zero(n):  # Here n is the formal parameter
    print(id(n))  # Address of n
    n = 0
    print(id(n))

a = 100  # a is the actual parameter
print(id(a))  # Address of a

zero(a)  # Passing a to n

print(a)
```

```python
def changeme(mylist1):
    print(id(mylist1))
    mylist1 = [1, 2, 3, 4]  # Assign a new reference to mylist1
    print(id(mylist1))
    print("Values inside the function:", mylist1)
    return

# Now you can call changeme function
mylist = [10, 20, 30]
print(id(mylist))

changeme(mylist)

print("Values outside the function:", mylist)
```
